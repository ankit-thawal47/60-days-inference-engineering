```python
from transformers import AutoTokenizer, AutoModel
import torch

# Load
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")
model = AutoModel.from_pretrained("distilbert-base-uncased")

# Text in
text = "I am learning inference engineering"
tokens = tokenizer(text, return_tensors="pt")  # pt = PyTorch tensors

print("input_ids shape:      ", tokens["input_ids"].shape)
print("attention_mask shape: ", tokens["attention_mask"].shape)

# Model forward pass
with torch.no_grad():
    output = model(**tokens)

print("Output shape: ", output.last_hidden_state.shape)
```
