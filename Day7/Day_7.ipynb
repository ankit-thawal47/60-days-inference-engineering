pipeline abstract away details of "inference"

1. tokenization
2. generate ids from the tokens
3. pass it through the model, so model takes input, applies the weights it learned, and outputs the output_ids
4. decode the actual words from the output_ids

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# Load
tokenizer = AutoTokenizer.from_pretrained("distilgpt2")
model = AutoModelForCausalLM.from_pretrained("distilgpt2")

# Step 1 — tokenize
prompt = "I am learning inference"
inputs = tokenizer(prompt, return_tensors="pt")
print("input_ids:", inputs["input_ids"])

# Step 2 — generate
output_ids = model.generate(
    inputs["input_ids"],
    max_new_tokens=20
)
print("output_ids:", output_ids)

# Step 3 — decode
output_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)
print("output_text:", output_text)
```
